```python
import torch
from diffusers import WanPipeline
from diffusers.utils import export_to_video

MODEL = "Wan-AI/Wan2.1-T2V-1.3B-Diffusers"
pipe = WanPipeline.from_pretrained(MODEL, torch_dtype=torch.bfloat16)
pipe.enable_model_cpu_offload()
pipe.vae.enable_tiling()
print("loaded", MODEL)
```

```python
prompt = ("9:16 vertical 2D motion graphics, pure white background, "
          "black minimalist stick figure round head no face, "
          "tiny village on seven hills, red arrows expanding upward, "
          "simple clean kinetic animation, no text, no words, no letters, no watermark")
negative = "photorealistic, 3d render, face, hair, clothes, gray background, shadows, text, words, letters, numbers, captions, subtitles, watermark, logo, blurry, extra limbs"
out = pipe(prompt=prompt, negative_prompt=negative, height=480, width=832, num_frames=81, guidance_scale=6.0, num_inference_steps=30).frames[0]
export_to_video(out, "clip1_rome.mp4", fps=16)
print("saved", len(out), "frames")
```
